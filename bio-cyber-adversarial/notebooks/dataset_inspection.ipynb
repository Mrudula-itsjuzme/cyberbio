# Dataset Inspection
Analyze the generated synthetic sequence dataset.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os

dataset_path = '../data/raw/dataset.csv'
df = pd.read_csv(dataset_path)
df.head()
```

```python
# Basic Information
print(f"Total samples: {len(df)}")
print("\nClass Distribution:")
print(df['label'].value_counts())
print("\nSplit Distribution:")
print(df['split'].value_counts())
```

```python
# Sequence length distribution
df['seq_length'] = df['sequence'].apply(len)
print("\nSequence Length Distribution:")
print(df['seq_length'].value_counts())

if df['seq_length'].nunique() == 1:
    print(f"\nAll sequences have length {df['seq_length'].iloc[0]}")
```

```python
# Validate characters
def check_invalid_chars(seq, valid_chars={'A', 'C', 'G', 'T'}):
    invalid = set(seq) - valid_chars
    return len(invalid) > 0

df['has_invalid_chars'] = df['sequence'].apply(check_invalid_chars)
invalid_count = df['has_invalid_chars'].sum()
print(f"\nSamples with invalid characters: {invalid_count}")
if invalid_count == 0:
    print("Success! All sequences contain only valid A, C, G, T characters.")
```

```python
# Example sequences
print("\nExample sequences from Class 0 (Motif: ACGTACGT):")
print(df[df['label'] == 0]['sequence'].head().values)

print("\nExample sequences from Class 1 (Motif: TGCATGCA):")
print(df[df['label'] == 1]['sequence'].head().values)
```
